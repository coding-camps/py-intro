## 1 数据读取

pandas需要先读取表格类型的数据，然后进行数据分析

| 数据类型           | 说明                | Pandas读取方法        |
|:---------------|:------------------|:------------------|
| csv, tsv, txt  | 用逗号分割，tab分割的纯文本文件 | pandas.read_csv   |
| excel          | 微软xls或xlsx格式      | pandas.read_excel |
| mysql          | 关系型数据库表           | pandas.read_sql   |

```python
csv_path = '~/datasets/ml-latest-small/ratings.csv'

# 读取数据
ratings = pd.read_csv(csv_path)
```

```python

# 查看前几行
rating_head = ratings.head()
print(f'rating_head:\n{rating_head}\n')

# 查看形状
rating_shape = ratings.shape
print(f'rating_shape:\n{rating_shape}\n')

# 查看列名属性
rating_cols = ratings.columns
print(f'rating_cols:\n{rating_cols}\n')

# 查看索引
rating_idx = ratings.index
print(f'rating_idx:\n{rating_idx}\n')
```
