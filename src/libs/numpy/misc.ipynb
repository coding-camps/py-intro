```python
import numpy as np


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


I = np.array([0.9, 0.1, 0.8]).reshape(3, 1)
W = np.array([
    [0.9, 0.3, 0.4],
    [0.2, 0.8, 0.2],
    [0.1, 0.5, 0.6]
])

X = np.dot(W, I)

Y = sigmoid(X)

print(I, I.shape)
print(W, W.shape)
print(X, X.shape)
print(Y, Y.shape)
```

### numpy.reshape()

不改变原形状。

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(f'a: {a.shape} = \n{a}')

b = np.reshape(a, (3, 2))
print(f'b: {b.shape} = \n{b}')

print(f'a: {a.shape} = \n{a}')

print(a.reshape(3, 2))
print(a.reshape(-1, 2))
print(a.reshape(3, -1))
```

### numpy.resize()
不改变原形状。

## 性能问题

### x = x + y vs x += y

```python
import numpy as np

x = np.arange(1, 10, 2)
print(x, type(x), id(x))

y = np.zeros_like(x)
print(y, type(y), id(y))

x = x + y
print(x, type(x), id(x))
```

```python
import numpy as np

x = np.arange(1, 10, 2)
print(x, type(x), id(x))

y = np.zeros_like(x)
print(y, type(y), id(y))

x += y
print(x, type(x), id(x))
```

## 广播机制

```python
import numpy as np

print('-' * 10)
a = np.arange(1, 13).reshape(2, 3, 2)
print(a.shape)
print(a)

print('-' * 10)
b = np.arange(1, 7).reshape(3, 2)
print(b.shape)
print(b)

print('-' * 10)
c = a + b
print(c.shape)
print(c)
```
