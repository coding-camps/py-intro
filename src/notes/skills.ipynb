```python
import random


def test_choice():
    li = range(0, 10)
    li = list(map(lambda ix: '[' + str(ix) + ']', li))
    # li = ['a', 'b', 'c', 'd', 'e', 'f']
    print(li)
    for i in range(1, 11):
        r = random.choice(li)
        print('%2s - %s' % (i, r), end='\n' if i % 5 == 0 else '  ')


test_choice()
```
