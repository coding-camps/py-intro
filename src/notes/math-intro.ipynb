### 求二次方程的根

求二次方程 $ax^2+bx+c = d (a \neq 0)$的根

```python
''' Quadratic equation root calculator '''


def quad_root(a, b, c, d):
    delta = float(b * b - 4 * a * c)
    if delta < 0:
        print("There is no real root but two complex roots:")
        root01 = (-b + delta ** 0.5) / (2 * a)
        root02 = (-b - delta ** 0.5) / (2 * a)
        print("{0}\n{1}\n".format(root01, root02))
    elif delta == 0.0:
        print("There is only one root:")
        root1 = -b / (2 * a)
        print("root is {}\n".format(root1))
    else:
        print("There are two real roots")
        root21 = (-b + delta ** 0.5) / (2 * a)
        root22 = (-b - delta ** 0.5) / (2 * a)
        print(f'{root21}\n{root22}\n')


quad_root(1, 1, 1, 1)
quad_root(1, 2, 1, 1)
quad_root(1, 500, -79, 10)
```

## 三、数据的统计学特征

- 均值 mean

```python
def cal_mean(data):
    total = sum(data)
    count = len(data)
    mean = total / count
    return mean


steps = [10000, 20000, 30000, 12500, 25000, 40000, 30000]
print(cal_mean(steps))
```

- 中位数 median

```python
def cal_median(data):
    nums = len(data)
    data.sort()
    if nums % 2 == 0:
        m1 = int(nums / 2) - 1
        m2 = int(nums / 2)
        return (data[m1] + data[m2]) / 2
    else:
        m = int(nums / 2)
        return data[m]


steps2 = [10000, 20000, 30000, 12500, 25000, 40000, 30000]
print('steps2 median : ', cal_median(steps2))
steps3 = steps2[0:-1]
print('steps3 median : ', cal_median(steps3))
```

- 众数 mode

```python
from collections import Counter


def cal_mode(data):
    counter = Counter(data)
    num_freq = counter.most_common()
    max_count = num_freq[0][1]
    modes = []
    for num_count in num_freq:
        if num_count[1] == max_count:
            modes.append(num_count[0])
    return modes


data1 = [1, 3, 2, 5, 5, 3]
print(cal_mode(data1))
data2 = [1, 5, 2, 3, 2, 3, 3]
print(cal_mode(data2))
```

- 极差 range

```python
def get_range(data):
    lowest = min(data)
    highest = max(data)
    range = highest - lowest
    return range


data1 = [1, 3, 2, 5, 5, 3]
print(get_range(data1))
```

- 方差 variance 标准差 standard deviation

```python
def cal_variance(data):
    total = sum(data)
    count = len(data)
    mean = total / count
    diff = [num - mean for num in data]
    squarde_diff = [ndiff ** 2 for ndiff in diff]
    variance = sum(squarde_diff) / count
    return variance


donations = [100, 60, 70, 900, 100, 200, 500, 500, 503, 600, 1000, 1200]
variance = cal_variance(donations)
print("variance =", variance)
print("standard deviation =", variance ** 0.5)

from matplotlib import pyplot as plt

plt.plot(range(len(donations)), donations, marker='+')
plt.show()
```

```python
# 解方程
from sympy import symbols, solve

x, y = symbols('x, y')
a, b, c = symbols('a, b, c')

## 解一元一次方程
expr = 5 * x - 10
result = solve(expr)
print(result)

## 解一元二次方程 # 结果中I表示复数
expr1 = x ** 2 + 5 * x + 4
result1 = solve(expr1)
print(result1)

expr2 = x ** 2 + x + 1
result2 = solve(expr2, dict=True)
print(result2)

## 解二元一次线性方程组
expr1 = 2 * x + 3 * y - 6
expr2 = 3 * x + 2 * y - 12
result = solve((expr1, expr2), dict=True)
print(result)
### 验证
er1 = expr1.subs({x: result[0][x], y: result[0][y]})
er2 = expr2.subs({x: result[0][x], y: result[0][y]})
print(er1, er2)

## 求解符号方程
expr = a * x ** 2 + b * x + c
result = solve(expr, x, dict=True)
print(result)

a11, a12, a21, a22 = symbols('a11, a12, a21, a22')
b1, b2 = symbols('b1, b2')
x1, x2 = symbols('x1, x2')
expr1 = a11 * x1 + a12 * x2 - b1
expr2 = a21 * x1 + a22 * x2 - b2
result = solve((expr1, expr2), (x1, x2), dict=True)
print(result)
```

```python
# 解不等式
## - 多项式 polynomial
from sympy import symbols, Symbol, Poly, solve_poly_inequality

x = Symbol('x')
# 不等式对象
ineq_obj = -x ** 2 + 4 < 0
# 提取不等式左侧
lhs = ineq_obj.lhs
# Poly指代上述提取的不等式
p = Poly(lhs, x)
# 提取关系运算符
rel = ineq_obj.rel_op
solve_poly_inequality(p, rel)
```

```python
# 解不等式
## - 有理式 rational expression
from sympy import symbols, Symbol, Poly, solve_rational_inequalities

x = Symbol('x')
# 创建有理不等式对象
ineq_obj = ((x - 1) / (x + 2)) > 0
lhs = ineq_obj.lhs
# 提取分子和分母
numer, denom = lhs.as_numer_denom()
# 多项式对象
p1 = Poly(numer)
p2 = Poly(denom)
# 提取关系运算符
rel = ineq_obj.rel_op
solve_rational_inequalities([[((p1, p2), rel)]])
```

```python
# 解不等式
## - 其他类型，不是多项式，也不是有理式
from sympy import symbols, Symbol, solve_univariate_inequality, sin

x = Symbol('x')
ineq_obj = sin(x) - 0.6 > 0
solve_univariate_inequality(ineq_obj, x, relational=False)
```

```python
import random
from math import pi


def get_area(radius, times):
    print(f'Radius: {radius}, Area: {pi * radius ** 2 : .5f}, ', end='')
    count = 0
    for t in range(times):
        x = random.random() * 2 * radius
        y = random.random() * 2 * radius
        if (x - radius) ** 2 + (y - radius) ** 2 <= radius ** 2:
            count += 1
    f = count / times
    print(f'Estimated({times} darts): area: {(2 * radius) ** 2 * f : .5f}, pi:{4 * f}')


if __name__ == '__main__':
    radius = 2
    times = [100, 1000, 100000, 100000, 1000000]
    for t in times:
        get_area(radius, t)
```

### 掷骰子最多max_rolls次，分数之和为 target_score，
### 输入这两个参数，求概率  

```python
from sympy import FiniteSet
import random


def find_prob(target_score, max_rolls):
    die_sides = FiniteSet(1, 2, 3, 4, 5, 6)
    if max_rolls > 6:
        return 0.0
    else:
        s = die_sides ** max_rolls
        success_rolls = []
        for elem in s:
            if sum(elem) >= target_score:
                success_rolls.append(elem)
        e = FiniteSet(*success_rolls)
        return len(e) / len(s)


if __name__ == '__main__':
    target_score, max_rolls = 25, 5
    p = find_prob(target_score, max_rolls)
    print(f'{p:.5f}')
```

### 分形0: 随机变换的案例

```python
from matplotlib import pyplot as plt
import random

plt.rcParams['axes.unicode_minus'] = False


def transformation_1(p):
    x = p[0]
    y = p[1]
    return x + 1, y - 1


def transformation_2(p):
    x = p[0]
    y = p[1]
    return x + 1, y + 1


def transform(p):
    transformations = [transformation_1, transformation_2]
    t = random.choice(transformations)
    x, y = t(p)
    return x, y


def build_trajectory(p, n):
    x = [p[0]]
    y = [p[1]]
    for i in range(n):
        p = transform(p)
        x.append(p[0])
        y.append(p[1])
    return x, y


if __name__ == '__main__':
    p = (1, 1)
    n = 100
    x, y = build_trajectory(p, n)
    plt.plot(x, y)
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.show()
```

```python
from matplotlib import pyplot as plt
import random

plt.rcParams['axes.unicode_minus'] = False


def transform_1(p):
    x = p[0]
    y = p[1]
    x1 = 0.85 * x + 0.04 * y
    y1 = -0.04 * x + 0.85 * y + 1.6
    return x1, y1


def transform_2(p):
    x = p[0]
    y = p[1]
    x1 = 0.2 * x - 0.26 * y
    y1 = 0.23 * x + 0.22 * y + 1.6
    return x1, y1


def transform_3(p):
    x = p[0]
    y = p[1]
    x1 = -0.15 * x + 0.28 * y
    y1 = 0.26 * x + 0.24 * y + 0.44
    return x1, y1


def transform_4(p):
    x = p[0]
    y = p[1]
    x1 = 0
    y1 = 0.16 * y
    return x1, y1


def get_index(prob):
    r = random.random()
    c_prob = 0
    sum_prob = []
    for p in prob:
        c_prob += p
        sum_prob.append(c_prob)
    for item, sp in enumerate(sum_prob):
        if r <= sp:
            return item
    return len(prob) - 1


def tranform(p):
    transformations = [transform_1, transform_2, transform_3, transform_4, ]
    probablity = [0.85, 0.07, 0.07, 0.01]
    tindex = get_index(probablity)
    t = transformations[tindex]
    x, y = t(p)
    return x, y


def draw_fern(n):
    x = [0]
    y = [0]

    x1, y1 = 0, 0
    for i in range(n):
        x1, y1 = tranform((x1, y1))
        x.append(x1)
        y.append(y1)
    return x, y


if __name__ == '__main__':
    n = 2000
    x, y = draw_fern(n)
    plt.plot(x, y, 'o')
    plt.title(f'Ferm with {n} points')
    plt.show()
```

### 分形2: Sierpinski三角形

```python
from matplotlib import pyplot as plt
import random

plt.rcParams['axes.unicode_minus'] = False


def transform_1(p):
    x = p[0]
    y = p[1]
    x1 = 0.5 * x
    y1 = 0.5 * y
    return x1, y1


def transform_2(p):
    x = p[0]
    y = p[1]
    x1 = 0.5 * x + 0.5
    y1 = 0.5 * y + 0.5
    return x1, y1


def transform_3(p):
    x = p[0]
    y = p[1]
    x1 = 0.5 * x + 1
    y1 = 0.5 * y
    return x1, y1


def get_index(prob):
    r = random.random()
    c_prob = 0
    sum_prob = []
    for p in prob:
        c_prob += p
        sum_prob.append(c_prob)
    for item, sp in enumerate(sum_prob):
        if r <= sp:
            return item
    return len(prob) - 1


def tranform(p):
    transformations = [transform_1, transform_2, transform_3, ]
    probablity = [1 / 3, 1 / 3, 1 / 3]
    tindex = get_index(probablity)
    t = transformations[tindex]
    x, y = t(p)
    return x, y


def draw_sierpinski(n):
    x = [0]
    y = [0]

    x1, y1 = 0, 0
    for i in range(n):
        x1, y1 = tranform((x1, y1))
        x.append(x1)
        y.append(y1)
    return x, y


if __name__ == '__main__':
    n = 5000
    x, y = draw_sierpinski(n)
    plt.plot(x, y, 'o')
    plt.title(f'Sierpinski with {n} points')
    plt.show()
```

```python
# 实际画图
# https://realpython.com/mandelbrot-set-python/
# https://www.cnblogs.com/anderslly/archive/2008/10/10/mandelbrot-set-by-fsharp.html
# https://matplotlib.org/matplotblog/posts/animated-fractals/
# https://matplotlib.org/stable/gallery/showcase/mandelbrot.html
# https://prettymathpics.com/mandelbrot-set-in-python-pygame/
# https://www.geeksforgeeks.org/mandelbrot-fractal-set-visualization-in-python/
# https://jhidding.github.io/modular-mandelbrot/
# https://matplotlib.org/2.0.2/examples/showcase/mandelbrot.html
# https://www.geeksforgeeks.org/mandelbrot-fractal-set-visualization-in-python/
# http://gael-varoquaux.info/scipy-lecture-notes/intro/numpy/auto_examples/plot_mandelbrot.html
# https://lectures.scientific-python.org/intro/numpy/auto_examples/plot_mandelbrot.html
# https://blog.51cto.com/u_15255081/5188041
# https://www.showapi.com/book/view/2125/25
# https://zhuanlan.zhihu.com/p/392574615
# https://blog.csdn.net/WhoisPo/article/details/49557823


# 时间逃逸算法
import matplotlib.pyplot as plt

# import matplotlib.cm as cm

plt.rcParams['axes.unicode_minus'] = False


def init_image(x_p, y_p):
    image = []
    for i in range(y_p):
        x_colors = []
        for j in range(x_p):
            x_colors.append(0)
        image.append(x_colors)
    return image


def mandelbrot(x, y, max_iteration):
    z = complex(0, 0)
    c = complex(x, y)
    iteration = 0
    while abs(z) < 2 and iteration < max_iteration:
        z = z ** 2 + c
        iteration += 1
    return iteration


def draw_set(point_start, point_end):
    x_p, y_p = 600, 600
    image = init_image(x_p, y_p)

    max_iteration = 500

    dx = (point_end[0] - point_start[0]) / (x_p - 1)
    dy = (point_end[1] - point_start[1]) / (y_p - 1)

    x_coords = [point_start[0] + dx * i for i in range(x_p)]
    y_coords = [point_start[1] + dy * i for i in range(y_p)]

    for i, x in enumerate(x_coords):
        for k, y in enumerate(y_coords):
            image[k][i] = mandelbrot(x, y, max_iteration)
            # image[i][j] = random.randint(0, 5)

    plt.figure(dpi=500)
    plt.imshow(
        image,
        # origin='lower',
        extent=(point_start[0], point_end[0], point_start[1], point_end[1]),
        # cmap=cm.Greys_r,
        # interpolation='nearest'
    )
    # plt.colorbar()
    # plt.savefig('m.jpg')
    plt.show()


if __name__ == '__main__':
    point_start = (-2.5, -1.0)
    point_end = (1.0, 1.0)
    draw_set(point_start, point_end)
```

### 函数极限

```python
from sympy import Limit, Symbol, S

x = Symbol('x')
expr = 1 / x
lim1 = Limit(expr, x, S.Infinity)
lim2 = Limit(expr, x, 0, dir='-')

print(lim1.doit())
print(lim2.doit())
```

### 函数求导

```python
from sympy import Symbol, Derivative

t = Symbol('t')
St = 5 * t ** 2 + 2 * t + 8
print(St)
d1 = Derivative(St, t)  # 默认1阶导数
print(d1.doit())
d2 = Derivative(St, t, 2)  # 2阶导数
print(d2.doit())

x = Symbol('x')
f = (x ** 3 + x ** 2 + x) * (x ** 2 + x)
d3 = Derivative(f, x)  # 自动使用求导规则
print(d3.doit())

y = Symbol('y')
g = 2 * x * y + x * y ** 2
d4 = Derivative(g, x)  # 对x求偏导数
d5 = Derivative(g, y)  # 对y求偏导数
print(d4.doit())
print(d5.doit())
```

### 求解积分

```python
# 不定积分： k * x
from sympy import Symbol, Integral

x = Symbol('x')
k = Symbol('k')
int1 = Integral(k * x, x)
int1.doit()
```

```python
# 定积分： x**2 在 0 到 4 之间曲边三角形面积
from sympy import Symbol, Integral

x = Symbol('x')
int1 = Integral(x ** 2, (x, 0, 4))
result = int1.doit()  # 计算
value = result.evalf()  # 数值结果
print(result, value)
```
