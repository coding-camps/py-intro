```python
# else - default
def test_for_else(use_continue: bool):
    print("------", "for-else", use_continue)
    for x in range(1, 11):
        if x > 8:
            if use_continue:
                continue
            else:
                break
        print(x, end=", ")
    else:
        print("[", x, "]", end=",")
    print()


if __name__ == '__main__':
    test_for_else(True)
    test_for_else(False)
```

```python
# for - else
def test_for_not():
    print("------", "for-not")
    for x in []:
        print("FOR-IN")
    else:
        print("FOR-ELSE")


if __name__ == '__main__':
    test_for_not()
```

```python
# while - else
def test_while_else(use_continue: bool):
    print("------", "while-else", use_continue)
    i = 0
    while i < 11:
        i += 1
        if i > 8:
            if use_continue:
                continue
            else:
                break
        print(i, end=', ')
    else:
        print("Flag", end='')
    print()


if __name__ == '__main__':
    test_while_else(True)
    test_while_else(False)
```

```python
# while - else
def test_while_not():
    print("------", "while-not")
    while False:
        print("WHILE-IN")
    else:
        print("WHILE-ELSE")


if __name__ == '__main__':
    test_while_not()
```

```python
# try - else
def test_else():
    print('test else ------')
    try:
        print('try')
    except Exception as e:
        print("catch:", e)
    else:
        print("else")
    finally:
        print("finally")


test_else()
```

```python
# Iterator

class Counter(object):

    def __init__(self, start=1, step=1):
        self.n = start - 1
        self.step = step

    def __iter__(self):
        return self

    def __next__(self):
        if self.n < 8:
            self.n += self.step
            return self.n
        else:
            raise StopIteration


def test_1():
    print("-" * 10, 1)
    c1 = Counter(1)
    c1_iter = iter(c1)
    for i in c1_iter:
        print(i, end=", ")
    print()


def test_2():
    print("-" * 10, 2)
    it = iter([1, 2, 3, 4, 5, 6])
    while True:
        try:
            x = next(it)
            print(x, end=" | ")
        except StopIteration as sie:
            print("exception: [", sie, ']')
            break


if __name__ == '__main__':
    test_1()
    test_2()
```

```python
# yield iterator, 携程（greenlet , gevent , asyncio）
# yield - 1
def my_iter():
    for i in range(1, 10):
        yield i
    return


def test1():
    print("-" * 10, 1)
    y_iter = my_iter()
    for i in y_iter:
        print(i, end=", ")
    print()


test1()
```

```python
# yield - 2
def fib1(n):
    n1 = 0
    n2 = 1
    initial = 0
    # n_current = 0
    while initial < n:
        # n_current = n1
        # print(n_current)
        print(n1, end=', ')
        n1, n2 = n2, n1 + n2
        initial += 1


def test_fib1():
    print('-' * 10, 'fib-1')
    fib1(8)
    print()


test_fib1()
```

```python
# Lambda - 3
def fib2(n):
    n1 = 0
    n2 = 1
    initial = 0
    # n_current = 0
    while initial < n:
        # n_current = n1
        # yield n_current
        yield n1
        n1, n2 = n2, n1 + n2
        initial += 1


def test_fib2_1():
    print('-' * 10, 'fib-2-1')
    f2 = fib2(8)
    for i in f2:
        print(i, end=', ')
    f2.close()
    print()


def test_fib2_2():
    print('-' * 10, 'fib-2-2')
    f2 = fib2(10)
    print(next(f2), end=" / ")
    print(next(f2), end=" / ")
    f2_send = f2.send("Hello")
    print(f2_send, end=" > ")
    print(next(f2), end=" / ")
    print(next(f2), end=" / ")
    f2.close()

test_fib2_1()
test_fib2_2()
```

```python
# Primer number
def test_prime(y):
    print("------", "prime")
    x = y // 2
    while x > 1:
        if y % x == 0:
            print(y, 'has factor', x)
            break
        x -= 1
    else:
        print(y, 'is prime')


if __name__ == '__main__':
    test_prime(100)
```
